### Bar Chart -:
1. It is used for Bivariate Analysis or Univariate Analysis.
2. Only in Numerical vs Categorical Columns.
3. Use case - Aggregate analysis of groups

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Simple bar chart -:
1. All previous chart function are applicable.
2. Color and Label Use as previous charts.

```python
#1. Vertical Bar Chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']
plt.bar(colors,children,color='skyblue')   # Colors = x-axix and children = y-axis
```

```python
#2. Horizontal Bar Chart
plt.barh(colors,children,color='skyblue')
```

### 2. Multiple Bar charts -:
There is no existing feature for make multiple bar chart in matplotlib but you can create by your logics and thinking.

```python
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='blue')
plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show()
```

### 3. xticks -:
Function used to get or set the current locations and lables of the tick marks on x-axis.

```python
# A problem without using xticks function
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']
plt.bar(colors,children,color='pink')
```

```python
# A Solution using xtick function
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']
plt.bar(colors,children,color='pink')
plt.xticks(rotation='vertical')
```

### 4. Stacked Bar chart -:
To create stacked bar chart you want to add upper side y-axis in bottom

```python
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')
plt.legend()
plt.show()
```
