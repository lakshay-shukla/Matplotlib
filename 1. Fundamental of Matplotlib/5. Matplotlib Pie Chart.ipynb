### Pie Chart -:
1. It is used for Univariate and Bivariate Analysis.
2. It's work on Categorical vs Numerical Data.
3. Use case - To find contribution on a standard scale.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Simple Data -:

```python
# Without Labels
data = [23,45,100,20,49]
plt.pie(data)
plt.show()
```

```python
# With Labels
data = [23,45,100,20,49]
subjects = ['eng','science','maths','sst','hindi']
plt.pie(data,labels=subjects)
plt.show()
```

### 2. On Some DataSet -:

```python
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
```

### 3. Percentage and Color -:

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

### 4. Explode Shadow -:

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.3,0,0,0,0,0.1],shadow=True)
plt.show()
```
