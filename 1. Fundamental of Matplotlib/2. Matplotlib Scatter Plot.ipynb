### Scatter Plot-:
1. It is used for Bivariate Analysis (Needs 2 columns to plot this graph).
2. Only on Numerical vs Numerical columns.
3. Use case - To Find correlation between 2 Quantities.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. plt.scatter simple function -:

```python
x = np.linspace(-10,10,50)
y = 10*x + 3 + np.random.randint(0,300,50)
y
```

```python
plt.scatter(x,y)
```

### 2. plt.scatter on pandas data -:

```python
df = pd.read_csv('batter.csv')
df = df.head(50)
df
```

```python
plt.scatter(df['avg'],df['strike_rate'],color='red')
plt.title('Avg and SR analysis of Top 50 Batsman')
plt.xlabel('Average')
plt.ylabel('SR')
```

```python
# marker -: used to highlight individual data points on line plots, scatter plots and error bars.
plt.scatter(df['avg'],df['strike_rate'],color='green',marker='+')
plt.title('Avg and SR analysis of Top 50 Batsman')
plt.xlabel('Average')
plt.ylabel('SR')
```

### 3. Size -:
It is slower them plt.plot

```python
tips = sns.load_dataset('tips')
# slower
plt.scatter(tips['total_bill'],tips['tip'],s=tips['size']*20)
```

### 4. scatterplot using plt.plot -:
It is faster then size.

```python
plt.plot(tips['total_bill'],tips['tip'],'o')
```
