```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### A Different ways/methods to plot a graph -:

#### 1. Method number 1 -:
Normal Method best for plotting single graph.

```python
batters = pd.read_csv('batter.csv')
batters.head()
```

```python
plt.figure(figsize=(15,6))
plt.scatter(batters['avg'],batters['strike_rate'])
plt.title('Avg and Strike Rate')
plt.xlabel('Avg')
plt.xlabel('Avg')
plt.ylabel('Strike Rate')
plt.show()
```

#### 2. Method number 2 -:
1. Subplot method for plotting multiple graphs.
2. fig = figure object ; ax = axis object ; figsize = ploting size(figure size).

```python
fig,ax = plt.subplots(figsize=(15,6))

ax.scatter(batters['avg'],batters['strike_rate'],color='red',marker='+')
ax.set_title('Avg and Strike Rate')
ax.set_xlabel('Avg')
ax.set_ylabel('Strike Rate')
```

#### 3. Method number 3 -:
1. First 2 = Rows; Second 2 = Columns; 1 = Which no. of graph.

```python
fig = plt.figure()
ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'])
```

### Subplots -:
1. It is used to print Multiple graphs side by side.

#### 1. Two Graph Side by Side -:

```python
fig, ax = plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(10,6))

ax[0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[1].scatter(batters['avg'],batters['runs'])

ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')

ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

#### 2. Three Graph Side by Side -:

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['avg'])
```

#### 3. Four Graph Side by Side -:

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='red')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'],color='green')
ax[1,1].hist(batters['runs'],color='yellow')
```
